## 1. Переменная **Usage_kWh**, float64

```python
X = df.features
X['Usage_kWh'].describe()
```

Значение присутствует для всех строк датасета.

```python
# Коэффициент вариации - данные неоднородны
print(X['Usage_kWh'].std() / X['Usage_kWh'].mean())

# Среднее и медиана. Большое смещение вправо
print(f"Среднее значение: {X['Usage_kWh'].mean()}")
print(f"Медиана: {X['Usage_kWh'].median()}")
```

```python
plt.figure(figsize=(10, 4))
plt.hist(X['Usage_kWh'], bins=100, edgecolor='black', alpha=0.7)
plt.title('Распределение Usage_kWh')
plt.xlabel('Потребление, кВт·ч')
plt.ylabel('Частота')
plt.grid(alpha=1)
plt.show()
```

```python
plt.figure(figsize=(10, 3))
plt.boxplot(X['Usage_kWh'], vert=False)
plt.title('Boxplot Usage_kWh')
plt.xlabel('кВт·ч')
plt.show()
```

```python
plt.figure(figsize=(10, 4))
positive = X['Usage_kWh'][X['Usage_kWh'] > 0]
plt.hist(np.log1p(positive), bins=100, edgecolor='black', alpha=0.7)
plt.title('log1p(Usage_kWh)')
plt.xlabel('log(1 + кВт·ч)')

plt.tight_layout()
plt.show()
```

```python
all_data = energy_data.data.features.join(energy_data.data.targets)

categories = ['Light_Load', 'Medium_Load', 'Maximum_Load']

colors = {
    'Light_Load':    '#2ca02c',   # зелёный
    'Medium_Load':   '#ffa600',   # желтый
    'Maximum_Load':  '#d62728',   # красный
}

plt.figure(figsize=(10, 6))

for i, cat in enumerate(categories):
    subset = all_data[all_data['Load_Type'] == cat]['Usage_kWh']
    x = i + np.random.normal(0, 0.2, size=len(subset))
    plt.scatter(x, subset, s=6, alpha=0.35, color=colors[cat], label=cat)

plt.xticks(range(len(categories)), categories)
plt.title('Все точки Usage_kWh по Load_Type на одном графике')
plt.xlabel('Load_Type')
plt.ylabel('кВт·ч')
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

Видим интересный момент, что Load_Type не задаёт жёсткий порог по кВт·ч. Завод может работать в режиме Medium_Load при 25 кВт·ч, а в другой день — при 55 кВт·ч. Режим — это качественная характеристика, а не точное число

```python
all_data.groupby('Load_Type')['Usage_kWh'].describe()
```

Данные по квантилям помогают лучше посмотреть перекрытие

В завершение посмотрим корреляцию со всеми остальными признаками

```python

numeric_df = all_data.select_dtypes(include=['number'])

# Матрица корреляций
corr_matrix = numeric_df.corr()
print(corr_matrix['Usage_kWh'].sort_values(ascending=False))

corr_target = corr_matrix['Usage_kWh'].drop('Usage_kWh').sort_values()

plt.figure(figsize=(8, 6))
colors = ['tab:red' if v > 0 else 'tab:blue' for v in corr_target.values]
plt.barh(corr_target.index, corr_target.values, color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('Корреляция Пирсона с Usage_kWh')
plt.title('Корреляция признаков с Usage_kWh')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```

## 2. Переменная **Lagging_Current_Reactive.Power_kVarh**, float64

Потребление реактивной мощности (кВАр·ч) в условиях отстающего тока.

Показатель Lagging_Current_Reactive.Power_kVarh отражает объем реактивной энергии (в киловар-часах), которая циркулирует между электросетью и индуктивным оборудованием (двигателями, трансформаторами, катушками) в условиях, когда фаза тока отстает от фазы напряжения. Физически эта энергия не расходуется на совершение полезной работы, а постоянно «перекачивается» туда и обратно для создания и поддержания электромагнитных полей, необходимых для запуска и функционирования промышленных машин, при этом дополнительно нагружая и нагревая провода.

```python
X['Lagging_Current_Reactive.Power_kVarh'].describe()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(X['Lagging_Current_Reactive.Power_kVarh'], bins=100, edgecolor='black', alpha=0.7)
axes[0].set_title('Исходное')
axes[0].set_xlabel('кВар·ч')

positive = X['Lagging_Current_Reactive.Power_kVarh'][X['Lagging_Current_Reactive.Power_kVarh'] > 0]
axes[1].hist(np.log1p(positive), bins=100, edgecolor='black', alpha=0.7)
axes[1].set_title('log1p, без нулей')
axes[1].set_xlabel('log(1 + кВар·ч)')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

sns.boxplot(data=all_data, x='Load_Type', y='Lagging_Current_Reactive.Power_kVarh', ax=axes[0])
axes[0].set_title('Lagging_Current_Reactive.Power_kVarh по Load_Type')
axes[0].tick_params(axis='x', rotation=30)

sns.boxplot(data=all_data, x='WeekStatus', y='Lagging_Current_Reactive.Power_kVarh', ax=axes[1])
axes[1].set_title('Lagging_Current_Reactive.Power_kVarh по WeekStatus')

plt.tight_layout()
plt.show()
```

```python
numeric_cols = all_data.select_dtypes(include=['number']).columns
corr_with_col = all_data[numeric_cols].corr()['Lagging_Current_Reactive.Power_kVarh'].sort_values(ascending=False)
print(corr_with_col)
```

## 3. Переменная **Lagging_Current_Power_Factor**, float64

Lagging_Current_Power_Factor — это коэффициент мощности при отстающем (индуктивном) токе, измеряемый в процентах (%). Этот признак описывает ту же физическую величину, что и Lagging_Current_Reactive.Power_kVarh (отстающая реактивная энергия), но с другой точки зрения. Реактивная энергия — это накопленная величина, а коэффициент мощности — это соотношение, отражающее эффективность нагрузки. Коэффициент мощности (Power Factor) является важным показателем эффективности электроэнергии: чем ближе значение к 1 (или 100%), тем эффективнее используется электроэнергия.

```python
X['Lagging_Current_Power_Factor'].describe()
```

```python
plt.figure(figsize=(10, 4))
plt.hist(X['Lagging_Current_Power_Factor'], bins=100, edgecolor='black', alpha=0.7)
plt.yscale('log')
plt.title('Распределение Lagging_Current_Power_Factor')
plt.xlabel('Коэффициент мощности, %')
plt.ylabel('Частота')
plt.grid(alpha=0.5)
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 4))

sns.boxplot(data=all_data, x='Load_Type', y='Lagging_Current_Power_Factor', ax=axes[0])
axes[0].set_title('Lagging_Current_Power_Factor')
axes[0].tick_params(axis='x', rotation=30)

sns.boxplot(data=all_data, x='WeekStatus', y='Lagging_Current_Power_Factor', ax=axes[1])
axes[1].set_title('Lagging_Current_Power_Factor')

plt.tight_layout()
plt.show()
```

```python
numeric_cols = all_data.select_dtypes(include=['number']).columns
corr_with_col = all_data[numeric_cols].corr()['Lagging_Current_Power_Factor'].sort_values(ascending=False)
print(corr_with_col)
```
